Es decir, fuera de $(-\pi,\pi)$ la señal se repite: $f(t+2\pi)=f(t)$. Su gráfica es una **onda diente de sierra** (una rampa que sube de $-\pi$ a $\pi$ y cae de golpe).

## 2. Introducción a la serie de Fourier

Una **serie de Fourier** representa una señal periódica como una **suma de senos y cosenos** (o, de forma equivalente, de exponenciales complejas) cuyas frecuencias son **múltiplos enteros de una frecuencia básica**. La idea central: cualquier señal periódica "razonable" se puede construir sumando ondas senoidales simples con la amplitud y la fase adecuadas.

### 2.1. Definiciones

| Concepto | Símbolo | Significado | En este ejercicio |
|---|---|---|---|
| **Período** | $T$ | Menor tiempo tras el cual la señal se repite: $f(t+T)=f(t)$ | $T=2\pi$ s |
| **Frecuencia fundamental** | $f_0=1/T$ | Cuántos períodos caben en un segundo, en Hz | $f_0=\dfrac{1}{2\pi}\approx0{,}159$ Hz |
| **Frecuencia angular fundamental** | $\omega_0=2\pi f_0$ | La misma frecuencia expresada en rad/s | $\omega_0=1$ rad/s |
| **Armónico $n$** | $n\omega_0$ | Onda cuya frecuencia es $n$ veces la fundamental ($n=1,2,3,\dots$). El armónico 1 es la fundamental | $n\cdot1$ rad/s |
| **Componente continua (DC)** | $a_0/2$ | Valor promedio de la señal; es la componente de frecuencia cero | se calcula en 3.1 |

La relación que enlaza período y frecuencia angular es

$$
\boxed{\omega_0=\frac{2\pi}{T}}
$$

Para el ejercicio, con $T=2\pi$:

$$
\omega_0=\frac{2\pi}{2\pi}=1\ \text{rad/s}
$$

Esto simplifica todo: los armónicos son $\cos(nt)$ y $\sin(nt)$ en lugar de $\cos(n\omega_0t)$ y $\sin(n\omega_0t)$. **Cuidado:** esto es una coincidencia de este ejercicio ($T=2\pi$); en general $\omega_0\neq1$.

```python
T = 2 * np.pi            # período, s
f0 = 1 / T               # frecuencia fundamental, Hz
w0 = 2 * np.pi / T       # frecuencia angular fundamental, rad/s

print(f"T  = {T:.4f} s")
print(f"f0 = {f0:.4f} Hz")
print(f"w0 = {w0:.4f} rad/s")
```

## 3. Resolución por el método trigonométrico

### 3.0. La forma de la serie y las fórmulas de los coeficientes

$$
\boxed{f(t)=\frac{a_0}{2}+\sum_{n=1}^{\infty}\Big[a_n\cos(n\omega_0t)+b_n\sin(n\omega_0t)\Big]}
$$

- $\dfrac{a_0}{2}$: componente continua (valor promedio).
- $a_n$: amplitud del **coseno** del armónico $n$.
- $b_n$: amplitud del **seno** del armónico $n$.

Los coeficientes se obtienen proyectando la señal sobre cada armónico, integrando en **un período completo** (aquí, de $-T/2$ a $T/2$):

$$
\boxed{a_0=\frac{2}{T}\int_{-T/2}^{T/2}f(t)\,dt}
\qquad
\boxed{a_n=\frac{2}{T}\int_{-T/2}^{T/2}f(t)\cos(n\omega_0t)\,dt}
\qquad
\boxed{b_n=\frac{2}{T}\int_{-T/2}^{T/2}f(t)\sin(n\omega_0t)\,dt}
$$

Con $T=2\pi$ el factor $2/T$ vale $1/\pi$ y los límites son $-\pi$ y $\pi$.

### 3.1. Cálculo de $a_0$

$$
a_0=\frac{1}{\pi}\int_{-\pi}^{\pi}t\,dt
=\frac{1}{\pi}\left[\frac{t^2}{2}\right]_{-\pi}^{\pi}
=\frac{1}{\pi}\left(\frac{\pi^2}{2}-\frac{(-\pi)^2}{2}\right)
=\frac{1}{\pi}\left(\frac{\pi^2}{2}-\frac{\pi^2}{2}\right)
$$

$$
\boxed{a_0=0}
$$

**Significado físico:** $a_0/2$ es el **valor promedio** de la señal en un período. La rampa recorre valores negativos y positivos de forma simétrica, así que el área por debajo del eje cancela exactamente la de arriba y el promedio es cero. La señal **no tiene componente continua (DC)**.

### 3.2. Cálculo de $a_n$ — uso de la paridad

Una función $g$ es **par** si $g(-t)=g(t)$ e **impar** si $g(-t)=-g(t)$. Para el producto de dos funciones valen las mismas reglas que para los signos:

| Producto | Resultado |
|---|---|
| par × par | par |
| impar × impar | par |
| par × impar | **impar** |

En $a_n$ el integrando es $t\cos(nt)$:

- $t$ es **impar** (porque $(-t)=-t$),
- $\cos(nt)$ es **par** (porque $\cos(-nt)=\cos(nt)$),
- por tanto $g(t)=t\cos(nt)$ es **impar**.

**Por qué la integral de una función impar sobre un intervalo simétrico es cero.** Partimos la integral en dos mitades y cambiamos $t\to-t$ en la primera:

$$
\int_{-\pi}^{\pi}g(t)\,dt=\int_{-\pi}^{0}g(t)\,dt+\int_{0}^{\pi}g(t)\,dt
$$

Con $u=-t$, $dt=-du$, y los límites $t=-\pi\to u=\pi$, $t=0\to u=0$:

$$
\int_{-\pi}^{0}g(t)\,dt=-\int_{\pi}^{0}g(-u)\,du=\int_{0}^{\pi}g(-u)\,du=-\int_{0}^{\pi}g(u)\,du
$$

(la última igualdad usa que $g$ es impar). Al sumar, las dos mitades se cancelan:

$$
\int_{-\pi}^{\pi}t\cos(nt)\,dt=0
\quad\Longrightarrow\quad
\boxed{a_n=\frac{1}{\pi}\int_{-\pi}^{\pi}t\cos(nt)\,dt=0}
$$

Esto vale para todo $n\geq1$: **ningún coseno aparece en la serie**. Es lo esperable: $f(t)=t$ es una función impar y los cosenos (pares) no pueden aportar a una función impar.

### 3.3. Cálculo de $b_n$

El integrando es ahora $t\sin(nt)$: $t$ es impar y $\sin(nt)$ es impar, así que el producto es **par**. Para una función par, las dos mitades de la integral son iguales y se suman:

$$
\int_{-\pi}^{\pi}t\sin(nt)\,dt=2\int_{0}^{\pi}t\sin(nt)\,dt
\quad\Longrightarrow\quad
b_n=\frac{1}{\pi}\cdot2\int_{0}^{\pi}t\sin(nt)\,dt=\boxed{\frac{2}{\pi}\int_{0}^{\pi}t\sin(nt)\,dt}
$$

**Integración por partes.** La regla es

$$
\int u\,dv=uv-\int v\,du
$$

Elegimos $u$ como la parte que se simplifica al derivar ($t$) y $dv$ como la que sabemos integrar ($\sin(nt)\,dt$):

$$
u=t\qquad\Rightarrow\qquad du=dt
$$

$$
dv=\sin(nt)\,dt\qquad\Rightarrow\qquad v=-\frac{\cos(nt)}{n}
$$

Sustituimos:

$$
\int t\sin(nt)\,dt=t\left(-\frac{\cos(nt)}{n}\right)-\int\left(-\frac{\cos(nt)}{n}\right)dt
=-\frac{t\cos(nt)}{n}+\frac{1}{n}\int\cos(nt)\,dt
$$

Como $\int\cos(nt)\,dt=\dfrac{\sin(nt)}{n}$:

$$
\int t\sin(nt)\,dt=-\frac{t\cos(nt)}{n}+\frac{\sin(nt)}{n^2}
$$

Entonces

$$
b_n=\frac{2}{\pi}\left[-\frac{t\cos(nt)}{n}+\frac{\sin(nt)}{n^2}\right]_0^\pi
$$

Evaluamos con $\cos(n\pi)=(-1)^n$ y $\sin(n\pi)=0$ (para $n$ entero):

- En $t=\pi$: $\;-\dfrac{\pi(-1)^n}{n}+\dfrac{0}{n^2}=-\dfrac{\pi(-1)^n}{n}$
- En $t=0$: $\;-\dfrac{0\cdot\cos 0}{n}+\dfrac{\sin 0}{n^2}=0$

$$
b_n=\frac{2}{\pi}\left(-\frac{\pi(-1)^n}{n}-0\right)=-\frac{2(-1)^n}{n}
$$

y como $-(-1)^n=(-1)^{n+1}$:

$$
\boxed{b_n=\frac{2(-1)^{n+1}}{n}}
$$

**Comprobación con un valor:** para $n=1$, $b_1=2(-1)^2/1=2$; para $n=2$, $b_2=2(-1)^3/2=-1$. Se verifica numéricamente más abajo.

```python
# Verificación numérica de a0, a_n y b_n con la integral aproximada por una suma
# (suma de Riemann con 200 000 puntos). No se usa ningún bucle: n va como columna.
tt = np.linspace(-np.pi, np.pi, 200_000, endpoint=False)
dt = tt[1] - tt[0]
n = np.arange(1, 6).reshape(-1, 1)          # n = 1..5 como columna

a0 = np.sum(tt) * dt / np.pi
an = np.sum(tt * np.cos(n * tt), axis=1) * dt / np.pi
bn = np.sum(tt * np.sin(n * tt), axis=1) * dt / np.pi
bn_teorico = 2 * (-1.0) ** (n.flatten() + 1) / n.flatten()

print("a0 numérico:", round(a0, 6), "  (teórico: 0)")
print("an numérico:", np.round(an, 6), "  (teórico: 0)")
print("bn numérico:", np.round(bn, 6))
print("bn teórico :", np.round(bn_teorico, 6))
```

### 3.4. Resultado trigonométrico

Con $a_0=0$, $a_n=0$ y $b_n=\dfrac{2(-1)^{n+1}}{n}$, y $\omega_0=1$:

$$
\boxed{f(t)=2\sum_{n=1}^{\infty}\frac{(-1)^{n+1}}{n}\sin(nt)}
$$

Los primeros términos ($n=1,\dots,5$), con $b_n$ calculado arriba:

| $n$ | $b_n$ | término |
|---|---|---|
| 1 | $2$ | $2\sin t$ |
| 2 | $-1$ | $-\sin 2t$ |
| 3 | $\tfrac23$ | $\tfrac23\sin 3t$ |
| 4 | $-\tfrac12$ | $-\tfrac12\sin 4t$ |
| 5 | $\tfrac25$ | $\tfrac25\sin 5t$ |

$$
f(t)=2\sin t-\sin 2t+\frac23\sin 3t-\frac12\sin 4t+\frac25\sin 5t-\cdots
$$

**Cómo se construye y qué representa cada término.** El primer término, $2\sin t$, es la **fundamental**: una senoidal con el mismo período que la señal (da la forma general de subida y bajada). Cada término siguiente es un **armónico** que oscila más rápido y con amplitud menor ($|b_n|=2/n$ decrece como $1/n$); su función es *corregir* la forma: el signo alterno $(-1)^{n+1}$ hace que unos armónicos sumen y otros resten, y entre todos "enderezan" la suma hasta parecerse a la rampa y a su caída brusca.

## 4. Resolución mediante exponenciales complejas

### 4.0. La forma compleja

$$
\boxed{f(t)=\sum_{n=-\infty}^{\infty}C_n\,e^{jn\omega_0t}}
$$

Aquí $C_n$ es un número **complejo** (el coeficiente del armónico $n$) y $e^{jn\omega_0t}$ es una exponencial compleja que gira con frecuencia $n\omega_0$. Nótese que $n$ recorre **todos los enteros**, positivos y negativos. En la sección 6 se explica qué significa eso.

Los coeficientes se calculan con

$$
\boxed{C_n=\frac{1}{T}\int_{-T/2}^{T/2}f(t)\,e^{-jn\omega_0t}\,dt}
$$

Con $T=2\pi$, $\omega_0=1$ y $f(t)=t$:

$$
\boxed{C_n=\frac{1}{2\pi}\int_{-\pi}^{\pi}t\,e^{-jnt}\,dt}
$$

**Aviso importante:** la antiderivada que obtendremos tiene $n$ en el denominador, por lo que **sirve solo para $n\neq0$**. El caso $n=0$ ($C_0$) se calcula aparte, en 4.3.

### 4.1. La integral exponencial

De nuevo, integración por partes:

$$
u=t\quad\Rightarrow\quad du=dt
\qquad\qquad
dv=e^{-jnt}\,dt\quad\Rightarrow\quad v=\frac{e^{-jnt}}{-jn}
$$

(La integral de $e^{at}$ es $e^{at}/a$; aquí $a=-jn$. El factor $j$ aparece porque la constante del exponente es imaginaria.)

$$
\int te^{-jnt}\,dt=t\cdot\frac{e^{-jnt}}{-jn}-\int\frac{e^{-jnt}}{-jn}\,dt
=-\frac{te^{-jnt}}{jn}+\frac{1}{jn}\int e^{-jnt}\,dt
$$

Y la integral que queda es $\int e^{-jnt}\,dt=\dfrac{e^{-jnt}}{-jn}$, así que

$$
\int te^{-jnt}\,dt=-\frac{te^{-jnt}}{jn}+\frac{1}{jn}\cdot\frac{e^{-jnt}}{-jn}
=-\frac{te^{-jnt}}{jn}+\frac{e^{-jnt}}{-j^2n^2}
$$

Como $j^2=-1$, el denominador $-j^2n^2=n^2$:

$$
\boxed{\int te^{-jnt}\,dt=-\frac{te^{-jnt}}{jn}+\frac{e^{-jnt}}{n^2}}
$$

**Origen de cada término:** el primero, $uv$, es el producto $t\cdot v$; el segundo es $-\int v\,du$, que al integrar otra vez una exponencial produce un segundo factor $1/(jn)$ y, junto con el primero, un $n^2$ en el denominador (de ahí que $j\cdot j=-1$ elimine la unidad imaginaria).

### 4.2. Evaluación de los límites

$$
C_n=\frac{1}{2\pi}\left[-\frac{te^{-jnt}}{jn}+\frac{e^{-jnt}}{n^2}\right]_{-\pi}^{\pi}
$$

Necesitamos $e^{-jn\pi}$ y $e^{jn\pi}$. Por Euler (sección 5.1):

$$
e^{\pm jn\pi}=\cos(n\pi)\pm j\sin(n\pi)=(-1)^n\pm j\cdot0=(-1)^n
\quad\Longrightarrow\quad
\boxed{e^{jn\pi}=e^{-jn\pi}=(-1)^n}
$$

Evaluamos cada término en los dos extremos.

**Primer término**, $-\dfrac{te^{-jnt}}{jn}$:

- en $t=\pi$: $\;-\dfrac{\pi(-1)^n}{jn}$
- en $t=-\pi$: $\;-\dfrac{(-\pi)(-1)^n}{jn}=+\dfrac{\pi(-1)^n}{jn}$

Su valor en la integral definida es (superior − inferior):

$$
-\frac{\pi(-1)^n}{jn}-\frac{\pi(-1)^n}{jn}=-\frac{2\pi(-1)^n}{jn}
$$

**Segundo término**, $\dfrac{e^{-jnt}}{n^2}$:

- en $t=\pi$: $\;\dfrac{(-1)^n}{n^2}$
- en $t=-\pi$: $\;\dfrac{(-1)^n}{n^2}$

Su valor es la diferencia: $\dfrac{(-1)^n}{n^2}-\dfrac{(-1)^n}{n^2}=0$. **Los términos con $1/n^2$ se cancelan** porque $e^{-jn\pi}=e^{jn\pi}$.

Queda:

$$
C_n=\frac{1}{2\pi}\left[-\frac{2\pi(-1)^n}{jn}\right]=-\frac{(-1)^n}{jn}
$$

Usamos $\dfrac1j=-j$ (porque $j\cdot(-j)=-j^2=1$):

$$
C_n=-(-1)^n\cdot\frac{1}{j}\cdot\frac1n=-(-1)^n\cdot(-j)\cdot\frac1n
$$

$$
\boxed{C_n=\frac{j(-1)^n}{n}\qquad(n\neq0)}
$$

### 4.3. El caso $n=0$

$$
C_0=\frac{1}{2\pi}\int_{-\pi}^{\pi}t\,dt=\frac{1}{2\pi}\left[\frac{t^2}{2}\right]_{-\pi}^{\pi}=\frac{1}{2\pi}\left(\frac{\pi^2}{2}-\frac{\pi^2}{2}\right)
$$

$$
\boxed{C_0=0}
$$

**¿Por qué no se reemplaza $n=0$ en $C_n=j(-1)^n/n$?** Porque esa fórmula se dedujo dividiendo entre $n$ (la antiderivada tiene $1/n$ y $1/n^2$) y para $n=0$ la expresión sería $\frac{j}{0}$, indefinida. Para $n=0$ la exponencial vale $e^{0}=1$ y la integral es otra (la de $t$, no la de $te^{-jnt}$ con $n\neq0$). $C_0$ es el valor promedio de la señal, y coincide con $a_0/2=0$.

### 4.4. Resultado complejo

$$
\boxed{f(t)=\sum_{\substack{n=-\infty\\n\neq0}}^{\infty}\frac{j(-1)^n}{n}\,e^{jnt}}
$$

El término $n=0$ no aparece porque $C_0=0$. La suma incluye armónicos de **frecuencia positiva** ($n>0$) y de **frecuencia negativa** ($n<0$).

```python
# Verificación numérica de C_n = j(-1)^n / n  (sin bucles: n como columna, n != 0)
n_c = np.array([-5, -4, -3, -2, -1, 1, 2, 3, 4, 5]).reshape(-1, 1)

Cn = np.sum(tt * np.exp(-1j * n_c * tt), axis=1) * dt / (2 * np.pi)
Cn_teorico = 1j * (-1.0) ** n_c.flatten() / n_c.flatten()
C0 = np.sum(tt) * dt / (2 * np.pi)

print("C0 numérico:", np.round(C0, 6))
print()
print(" n | C_n numérico      | C_n teórico")
print("---+-------------------+------------------")
print(np.column_stack([n_c.flatten(), np.round(Cn, 4), np.round(Cn_teorico, 4)]))
```

## 5. Demostración de la equivalencia entre ambos métodos

No basta con ver que los dos resultados "se parecen": se demuestra de dónde sale la relación entre $(a_n,b_n)$ y $C_n$.

### 5.1. Identidades de Euler

Para cualquier ángulo $\theta$:

$$
e^{j\theta}=\cos\theta+j\sin\theta
\qquad\qquad
e^{-j\theta}=\cos\theta-j\sin\theta
$$

(la segunda sale de la primera con $\theta\to-\theta$, usando que $\cos$ es par y $\sin$ es impar).

**Sumando** las dos ecuaciones, el seno se cancela:

$$
e^{j\theta}+e^{-j\theta}=2\cos\theta
\quad\Longrightarrow\quad
\boxed{\cos\theta=\frac{e^{j\theta}+e^{-j\theta}}{2}}
$$

**Restando** (primera menos segunda), el coseno se cancela:

$$
e^{j\theta}-e^{-j\theta}=2j\sin\theta
\quad\Longrightarrow\quad
\boxed{\sin\theta=\frac{e^{j\theta}-e^{-j\theta}}{2j}}
$$

### 5.2. De la serie trigonométrica a la compleja

Tomamos un término genérico de la serie trigonométrica y sustituimos $\theta=n\omega_0t$:

$$
\begin{aligned}
a_n\cos(n\omega_0t)+b_n\sin(n\omega_0t)
&=a_n\frac{e^{jn\omega_0t}+e^{-jn\omega_0t}}{2}+b_n\frac{e^{jn\omega_0t}-e^{-jn\omega_0t}}{2j}
\end{aligned}
$$

Distribuimos y agrupamos, por separado, las exponenciales de exponente positivo y negativo:

$$
=\left(\frac{a_n}{2}+\frac{b_n}{2j}\right)e^{jn\omega_0t}+\left(\frac{a_n}{2}-\frac{b_n}{2j}\right)e^{-jn\omega_0t}
$$

Con $\dfrac1j=-j$, es decir $\dfrac{b_n}{2j}=-\dfrac{jb_n}{2}$:

$$
=\underbrace{\frac{a_n-jb_n}{2}}_{C_n}\,e^{jn\omega_0t}+\underbrace{\frac{a_n+jb_n}{2}}_{C_{-n}}\,e^{-jn\omega_0t}
$$

El primer paréntesis multiplica a la exponencial de frecuencia $+n\omega_0$ y es, por definición de la serie compleja, $C_n$. El segundo multiplica a la de frecuencia $-n\omega_0$, así que es $C_{-n}$:

$$
\boxed{C_n=\frac{a_n-jb_n}{2}}
\qquad\qquad
\boxed{C_{-n}=\frac{a_n+jb_n}{2}}
$$

Además, el término constante: $\dfrac{a_0}{2}=C_0$.

**¿Por qué la frecuencia positiva lleva el signo menos y la negativa el signo más?** Porque el coseno aporta $+\tfrac12$ a *ambas* exponenciales (es par), mientras que el seno aporta $+\tfrac{1}{2j}$ a la positiva y $-\tfrac{1}{2j}$ a la negativa (es impar). Al escribir $\tfrac1{2j}=-\tfrac j2$, el $+\tfrac{1}{2j}$ de la frecuencia positiva se vuelve $-\tfrac{j}{2}b_n$, y el $-\tfrac1{2j}$ de la negativa se vuelve $+\tfrac j2b_n$. El signo viene de la **imparidad del seno**.

### 5.3. La relación inversa

Partimos de

$$
C_n=\frac{a_n-jb_n}{2}\qquad C_{-n}=\frac{a_n+jb_n}{2}
$$

**Sumamos:** $\;C_n+C_{-n}=\dfrac{a_n-jb_n+a_n+jb_n}{2}=\dfrac{2a_n}{2}$

$$
\boxed{a_n=C_n+C_{-n}}
$$

**Restamos:** $\;C_n-C_{-n}=\dfrac{a_n-jb_n-a_n-jb_n}{2}=\dfrac{-2jb_n}{2}=-jb_n$

Despejando $b_n$ y usando $\dfrac{1}{-j}=j$:

$$
\boxed{b_n=j\,(C_n-C_{-n})}
$$

Estas fórmulas permiten volver de la representación compleja a la trigonométrica.

### 5.4. Aplicación al ejercicio

Con $a_n=0$ y $b_n=\dfrac{2(-1)^{n+1}}{n}$, reemplazamos en $C_n=\dfrac{a_n-jb_n}{2}$:

$$
C_n=\frac{0-j\left(\dfrac{2(-1)^{n+1}}{n}\right)}{2}=-\frac{j(-1)^{n+1}}{n}
$$

Como $(-1)^{n+1}=-(-1)^n$:

$$
C_n=-\frac{j\cdot\big(-(-1)^n\big)}{n}=\boxed{\frac{j(-1)^n}{n}}
$$

Es **idéntico** al resultado de la integración exponencial directa (sección 4.2). Los dos métodos producen la misma serie.

### 5.5. Verificación de que ambas series son la misma señal

Agrupamos en la serie compleja los términos $n$ y $-n$ (con $n\geq1$). Sabemos $C_n=\dfrac{j(-1)^n}{n}$ y $C_{-n}=\dfrac{j(-1)^{-n}}{-n}=-\dfrac{j(-1)^n}{n}$, pues $(-1)^{-n}=(-1)^n$:

$$
C_ne^{jnt}+C_{-n}e^{-jnt}=\frac{j(-1)^n}{n}\left(e^{jnt}-e^{-jnt}\right)
$$

Por la identidad del seno, $e^{jnt}-e^{-jnt}=2j\sin(nt)$:

$$
=\frac{j(-1)^n}{n}\cdot2j\sin(nt)=\frac{2j^2(-1)^n}{n}\sin(nt)=\frac{-2(-1)^n}{n}\sin(nt)=\frac{2(-1)^{n+1}}{n}\sin(nt)
$$

Esto es exactamente $b_n\sin(nt)$. Sumando sobre todos los $n\geq1$ se recupera la serie trigonométrica:

$$
\sum_{n\neq0}C_ne^{jnt}=\sum_{n=1}^{\infty}\Big(C_ne^{jnt}+C_{-n}e^{-jnt}\Big)=2\sum_{n=1}^{\infty}\frac{(-1)^{n+1}}{n}\sin(nt)\;\;\blacksquare
$$

Las dos series representan la misma señal. La comprobación numérica:

```python
# Reconstruye la señal con N armónicos por los DOS caminos y compara.
def serie_trigonometrica(t, N):
    n = np.arange(1, N + 1).reshape(-1, 1)
    bn = 2 * (-1.0) ** (n + 1) / n
    return np.sum(bn * np.sin(n * t), axis=0)

def serie_compleja(t, N):
    n = np.concatenate([np.arange(-N, 0), np.arange(1, N + 1)]).reshape(-1, 1)
    Cn = 1j * (-1.0) ** n / n
    return np.sum(Cn * np.exp(1j * n * t), axis=0)

t = np.linspace(-np.pi + 0.1, np.pi - 0.1, 1000)
f_trig = serie_trigonometrica(t, N=50)
f_comp = serie_compleja(t, N=50)

print("Máxima parte imaginaria de la serie compleja :", np.max(np.abs(np.imag(f_comp))))
print("Máxima diferencia |trigonométrica - compleja|:", np.max(np.abs(f_trig - np.real(f_comp))))
```

## 6. Frecuencias positivas y negativas

En la serie compleja

$$
f(t)=\sum_{n=-\infty}^{\infty}C_ne^{jn\omega_0t}
$$

el índice $n$ indica **qué armónico** es y **en qué sentido gira**:

| Índice | Exponencial | Frecuencia | Significado |
|---|---|---|---|
| $n>0$ | $e^{+jn\omega_0t}$ | $+n\omega_0$ | gira en sentido **antihorario** |
| $n<0$ | $e^{-j\lvert n\rvert\omega_0t}$ | $-\lvert n\rvert\omega_0$ | gira en sentido **horario** |
| $n=0$ | $e^{0}=1$ | $0$ | no gira: es la componente **continua** ($C_0$) |

### 6.1. $e^{j\omega t}$ frente a $e^{-j\omega t}$

Por Euler:

$$
e^{j\omega t}=\cos(\omega t)+j\sin(\omega t)\qquad\qquad e^{-j\omega t}=\cos(\omega t)-j\sin(\omega t)
$$

| | $e^{j\omega t}$ | $e^{-j\omega t}$ |
|---|---|---|
| Parte real | $\cos(\omega t)$ | $\cos(\omega t)$ |
| Parte imaginaria | $+\sin(\omega t)$ | $-\sin(\omega t)$ |
| Sentido de giro en el plano complejo | antihorario | horario |

Ambas tienen la **misma parte real** y **partes imaginarias opuestas**: una es la **conjugada** de la otra, $e^{-j\omega t}=(e^{j\omega t})^*$. Son dos puntos que giran sobre la misma circunferencia unitaria en sentidos contrarios y siempre simétricos respecto al eje real.

**Las frecuencias negativas no son frecuencias físicas negativas.** No existe un oscilador con "−3 Hz". Son una consecuencia de la notación: para escribir un coseno o un seno reales con exponenciales complejas hacen falta **dos** de ellas, una girando a cada lado, de modo que sus partes imaginarias se cancelen y la suma sea real. La frecuencia negativa es el costo de usar la exponencial compleja, que es mucho más cómoda para operar que senos y cosenos.

### 6.2. Simetría conjugada para señales reales

Si $f(t)$ es **real**, entonces $f(t)=f(t)^*$. Conjugamos la serie compleja (el conjugado de una suma es la suma de los conjugados y $(e^{jn\omega_0t})^*=e^{-jn\omega_0t}$):

$$
f(t)=\sum_{n}C_ne^{jn\omega_0t}\;\Rightarrow\;f(t)^*=\sum_{n}C_n^*\,e^{-jn\omega_0t}=\sum_{m}C_{-m}^*\,e^{jm\omega_0t}
$$

(en la última igualdad se renombra $m=-n$). Para que $f(t)=f(t)^*$, los coeficientes de $e^{jm\omega_0t}$ deben coincidir: $C_m=C_{-m}^*$, es decir

$$
\boxed{C_{-n}=C_n^*}
$$

El asterisco indica el **conjugado complejo** (cambiar $j$ por $-j$). También se ve en 5.2: $C_n=\dfrac{a_n-jb_n}{2}$ y $C_{-n}=\dfrac{a_n+jb_n}{2}$ son conjugados porque $a_n$ y $b_n$ son reales.

**Ejemplos de este ejercicio** con $C_n=\dfrac{j(-1)^n}{n}$:

| $n$ | $C_n$ | $C_{-n}$ | ¿$C_{-n}=C_n^*$? |
|---|---|---|---|
| 1 | $\dfrac{j(-1)}{1}=-j$ | $j$ | sí: $(-j)^*=j$ |
| 2 | $\dfrac{j(+1)}{2}=\dfrac j2$ | $-\dfrac j2$ | sí: $\left(\dfrac j2\right)^*=-\dfrac j2$ |

Los pares $C_ne^{jnt}$ y $C_{-n}e^{-jnt}$ son **conjugados entre sí**, y la suma de un número complejo y su conjugado es el doble de su parte real: $z+z^*=2\operatorname{Re}\{z\}$. Por eso cada par produce una onda **real**:

$$
C_ne^{jnt}+C_{-n}e^{-jnt}=2\operatorname{Re}\{C_ne^{jnt}\}
$$

(Compruébalo con $n=1$, $C_1=-j$: $2\operatorname{Re}\{-j(\cos t+j\sin t)\}=2\operatorname{Re}\{-j\cos t+\sin t\}=2\sin t$. Ese es el primer armónico.)

## 7. Reconstrucción del seno

Tomemos $f(t)=\sin(\omega_0t)$. Por la identidad de Euler del seno:

$$
\sin(\omega_0t)=\frac{e^{j\omega_0t}-e^{-j\omega_0t}}{2j}
$$

Separamos cada exponencial y usamos $\dfrac1{2j}=-\dfrac j2$:

$$
\sin(\omega_0t)=\frac{1}{2j}e^{j\omega_0t}-\frac{1}{2j}e^{-j\omega_0t}
=\boxed{-\frac{j}{2}\,e^{j\omega_0t}+\frac{j}{2}\,e^{-j\omega_0t}}
$$

Comparando con $f(t)=\sum C_ne^{jn\omega_0t}$:

$$
C_1=-\frac j2\qquad C_{-1}=\frac j2\qquad C_n=0\ \text{para el resto}
$$

Un seno **necesita ambas frecuencias** ($+\omega_0$ y $-\omega_0$): cada exponencial por separado es compleja, y solo la combinación con coeficientes conjugados ($-\tfrac j2$ y $+\tfrac j2$) tiene parte imaginaria nula. Los coeficientes son **imaginarios puros** y de **signo opuesto** (antisimétricos): es la huella de una función **impar**.

Esto concuerda con el ejercicio: el primer armónico de $f(t)=t$ es $2\sin t$, con $C_1=2\cdot\left(-\tfrac j2\right)=-j$.

---
## 8. Reconstrucción del coseno

Tomemos $f(t)=\cos(\omega_0t)$:

$$
\cos(\omega_0t)=\frac{e^{j\omega_0t}+e^{-j\omega_0t}}{2}=\boxed{\frac12e^{j\omega_0t}+\frac12e^{-j\omega_0t}}
$$

$$
C_1=\frac12\qquad C_{-1}=\frac12
$$

Los coeficientes son **reales** e **iguales** (simétricos). Son conjugados ($\tfrac12$ es su propio conjugado) y la suma es real porque ambos tienen la misma parte real y partes imaginarias que se cancelan. Es la huella de una función **par**.

**Resumen:** la señal par (coseno) da coeficientes reales y simétricos, $C_{-n}=C_n$; la impar (seno) da coeficientes imaginarios y antisimétricos, $C_{-n}=-C_n$. Nuestro $f(t)=t$ es impar, y por eso sus $C_n$ son imaginarios puros con $C_{-n}=-C_n$.

### 9.2. Armónicos negativos

$C_{-n}=\dfrac{a_n+jb_n}{2}=\dfrac{j\,b_n}{2}=-C_n$ (aquí, porque $a_n=0$). Equivale a $C_{-n}=C_n^*$.

| $n$ | $C_n$ |
|---|---|
| $-1$ | $j$ |
| $-2$ | $-\dfrac j2$ |
| $-3$ | $\dfrac j3$ |
| $-4$ | $-\dfrac j4$ |
| $-5$ | $\dfrac j5$ |

**Simetría conjugada:** cada fila negativa es el conjugado de la fila positiva con el mismo $|n|$: $C_{-1}=j=(-j)^*=C_1^*$, $C_{-2}=-\tfrac j2=\left(\tfrac j2\right)^*=C_2^*$, etc. Como los $C_n$ son imaginarios puros, conjugar equivale a cambiar de signo. Además, la magnitud $|C_n|=\dfrac1{|n|}$ es la misma a ambos lados: **el espectro de magnitud es par; el de fase es impar**.

```python
# Tabla numérica de los coeficientes (sin bucles: todo vectorizado).
n_t = np.arange(1, 6)
a_n = np.zeros(5)
b_n = 2 * (-1.0) ** (n_t + 1) / n_t
C_n = (a_n - 1j * b_n) / 2                 # relación del apartado 5.2
C_directo = 1j * (-1.0) ** n_t / n_t       # fórmula obtenida en el apartado 4.2

print("  n |   a_n |    b_n  |    C_n     | C_n directo | conjugado = C_-n")
print(np.column_stack([n_t, a_n, np.round(b_n, 4), np.round(C_n, 4),
                       np.round(C_directo, 4), np.round(np.conj(C_n), 4)]))

print()
print("¿Coinciden las dos rutas para C_n?   ", np.allclose(C_n, C_directo))
print("¿Se cumple C_-n = conj(C_n)?         ", np.allclose(-C_n, np.conj(C_n)))
```

## 10. Aproximación con los primeros armónicos

Con cinco armónicos:

$$
f_5(t)=2\sin t-\sin 2t+\frac23\sin 3t-\frac12\sin 4t+\frac25\sin 5t
$$

En general, la **suma parcial** con $N$ armónicos es

$$
f_N(t)=2\sum_{n=1}^{N}\frac{(-1)^{n+1}}{n}\sin(nt)
$$

**Qué ocurre al aumentar $N$.** Cada armónico añadido mejora el ajuste: la aproximación se parece cada vez más a la rampa, y el error medio disminuye. Los armónicos de orden alto tienen amplitud $2/n$ pequeña, por lo que sólo corrigen detalles finos y los saltos.

**Convergencia.**

- En los puntos donde $f$ es **continua** ($t\neq\pm\pi,\pm3\pi,\dots$), $f_N(t)\to f(t)$ cuando $N\to\infty$.
- En los **saltos** (aquí en $t=\pm\pi$, donde la señal pasa de $+\pi$ a $-\pi$), la serie converge al **promedio de los límites laterales**:

$$
f(\pi)=\frac{f(\pi^-)+f(\pi^+)}{2}=\frac{\pi+(-\pi)}{2}=0
$$

Se comprueba directamente: en $t=\pi$ todos los términos $\sin(n\pi)=0$, así que la serie vale $0$.

**Fenómeno de Gibbs.** Cerca de un salto, las sumas parciales **sobrepasan** el valor de la señal formando una oscilación: el pico alcanza alrededor de un **9 %** del tamaño del salto (aquí, del salto de $2\pi$) y **no desaparece al aumentar $N$**; solo se vuelve más angosto y se pega al punto de discontinuidad. Es la consecuencia de intentar construir un salto abrupto con funciones suaves.

## 11. Código Python

Cambia `N_ARMONICOS` para ver el efecto de añadir armónicos.

```python
# ----------------------------------------------------------------------
# Serie de Fourier de f(t) = t en (-pi, pi), período T = 2 pi
# ----------------------------------------------------------------------
N_ARMONICOS = 5      # <-- modifica este valor: 1, 3, 5, 10, 50 ...

def senal_original(t):
    """f(t) = t en (-pi, pi), repetida con período 2*pi (diente de sierra)."""
    return np.mod(t + np.pi, 2 * np.pi) - np.pi

def aproximacion(t, N):
    """Suma parcial f_N(t) = 2 * sum_{n=1}^{N} (-1)^(n+1)/n * sin(n t)."""
    n = np.arange(1, N + 1).reshape(-1, 1)     # columna n = 1..N
    bn = 2 * (-1.0) ** (n + 1) / n             # coeficientes b_n
    return np.sum(bn * np.sin(n * t), axis=0)  # suma sobre los armónicos

def graficar(N):
    """Dibuja la señal original y su aproximación con N armónicos."""
    t = np.linspace(-3 * np.pi, 3 * np.pi, 4000)
    plt.plot(t, senal_original(t), label="f(t) original")
    plt.plot(t, aproximacion(t, N), label=f"Serie con N = {N} armónicos")
    plt.title(f"Serie de Fourier de f(t) = t con N = {N} armónicos")
    plt.xlabel("Tiempo t [s]")
    plt.ylabel("Amplitud")
    plt.legend()

plt.figure(figsize=(10, 4.5))
graficar(N_ARMONICOS)
plt.show()
```

```python
# Comparación de varios valores de N en una sola figura
plt.figure(figsize=(11, 9))

plt.subplot(2, 2, 1); graficar(1)
plt.subplot(2, 2, 2); graficar(3)
plt.subplot(2, 2, 3); graficar(10)
plt.subplot(2, 2, 4); graficar(50)

plt.tight_layout()
plt.show()
```

```python
# Fenómeno de Gibbs: el sobrepico cerca del salto no desaparece al aumentar N.
# (Con N pequeño el pico todavía no se ha formado; crece hacia ~9 % al aumentar N.)
t_cerca = np.linspace(0.0, np.pi, 20_000)       # tramo previo al salto en t = pi
salto = 2 * np.pi                               # f pasa de +pi a -pi

def sobrepico(N):
    """Máximo de f_N en [0, pi] menos el valor pi de la señal, en % del salto."""
    return (np.max(aproximacion(t_cerca, N)) - np.pi) / salto * 100

print(f"N =   5 -> sobrepico = {sobrepico(5):.2f} % del salto")
print(f"N =  20 -> sobrepico = {sobrepico(20):.2f} % del salto")
print(f"N = 100 -> sobrepico = {sobrepico(100):.2f} % del salto")
print(f"N = 500 -> sobrepico = {sobrepico(500):.2f} % del salto")

# En el propio salto la serie vale el promedio de los límites laterales
print()
print("f_N(pi) con N = 50 :", np.round(aproximacion(np.array([np.pi]), 50)[0], 10),
      "(valor teórico 0)")
```

## 12. Conclusiones

- **Método trigonométrico.** Como $f(t)=t$ es impar, $a_0=0$ y $a_n=0$ por paridad; solo sobreviven los senos, con $b_n=\dfrac{2(-1)^{n+1}}{n}$. El cálculo exige integración por partes en el semiperíodo $[0,\pi]$.
- **Método complejo.** Una sola integral, $C_n=\dfrac1{2\pi}\int_{-\pi}^{\pi}te^{-jnt}dt$, entrega $C_n=\dfrac{j(-1)^n}{n}$ para $n\neq0$, y $C_0=0$ se calcula aparte. El mismo procedimiento sirve para cualquier $n$, positivo o negativo.
- **Por qué coinciden.** Las identidades de Euler convierten cada $\cos$ y $\sin$ en dos exponenciales; al agrupar queda $C_n=\dfrac{a_n-jb_n}{2}$ y $C_{-n}=\dfrac{a_n+jb_n}{2}$. Son dos escrituras de la misma suma. Con $a_n=0$ y $b_n=\dfrac{2(-1)^{n+1}}{n}$ se recupera $C_n=\dfrac{j(-1)^n}{n}$.
- **Relación de coeficientes.** $a_n=C_n+C_{-n}$ y $b_n=j(C_n-C_{-n})$, además de $C_0=a_0/2$.
- **Frecuencias positivas y negativas.** Son exponenciales que giran en sentidos opuestos; no son frecuencias físicas negativas, sino la forma de representar un seno o un coseno real como suma de dos exponenciales conjugadas.
- **Simetría conjugada.** Si la señal es real, $C_{-n}=C_n^*$: es lo que garantiza que las partes imaginarias se cancelen al sumar cada par.
- **Cuándo usar cada una.** La forma **trigonométrica** es más intuitiva y conviene cuando se quiere ver amplitudes de senos y cosenos o al trabajar con paridad. La **compleja** es más compacta, trata todos los armónicos con una sola fórmula y es la que se generaliza a la transformada de Fourier; es la preferida para análisis de sistemas.

## Apéndice. Identidades útiles

Este apéndice reúne las identidades que se usan (o que conviene tener a mano) al trabajar con series de Fourier.

### A.1. Números complejos y potencias de $j$

| Identidad | Expresión |
|---|---|
| Unidad imaginaria | $j^2=-1$ |
| Inverso de $j$ | $\dfrac1j=-j$ |
| Potencias de $j$ | $j^0=1,\;j^1=j,\;j^2=-1,\;j^3=-j,\;j^4=1$ (se repite cada 4) |
| Conjugado | $(a+jb)^*=a-jb$ |
| Módulo | $\lvert a+jb\rvert=\sqrt{a^2+b^2}=\sqrt{zz^*}$ |
| Parte real e imaginaria | $\operatorname{Re}\{z\}=\dfrac{z+z^*}{2}$, $\;\operatorname{Im}\{z\}=\dfrac{z-z^*}{2j}$ |
| Conjugado de un producto | $(z_1z_2)^*=z_1^*z_2^*$ |
| Forma polar | $z=\lvert z\rvert e^{j\angle z}$ |

### A.2. Euler y exponenciales complejas

| Identidad | Expresión |
|---|---|
| Euler | $e^{j\theta}=\cos\theta+j\sin\theta$ |
| Conjugada | $(e^{j\theta})^*=e^{-j\theta}$ |
| Módulo unitario | $\lvert e^{j\theta}\rvert=1$ |
| Producto | $e^{j\alpha}e^{j\beta}=e^{j(\alpha+\beta)}$ |
| De Moivre | $(e^{j\theta})^n=e^{jn\theta}=\cos(n\theta)+j\sin(n\theta)$ |
| Valores notables | $e^{j0}=1,\;e^{j\pi/2}=j,\;e^{j\pi}=-1,\;e^{j3\pi/2}=-j,\;e^{j2\pi}=1$ |
| Periodicidad | $e^{j(\theta+2\pi)}=e^{j\theta}$ |
| Para $n$ entero | $e^{jn\pi}=e^{-jn\pi}=(-1)^n$, $\;e^{jn2\pi}=1$ |

### A.3. Trigonometría

| Identidad | Expresión |
|---|---|
| Pitágoras | $\sin^2\theta+\cos^2\theta=1$ |
| Paridad | $\cos(-\theta)=\cos\theta$, $\;\sin(-\theta)=-\sin\theta$ |
| Desfase de $\pi/2$ | $\sin\theta=\cos\left(\theta-\tfrac\pi2\right)$ |
| Suma de ángulos (seno) | $\sin(\alpha\pm\beta)=\sin\alpha\cos\beta\pm\cos\alpha\sin\beta$ |
| Suma de ángulos (coseno) | $\cos(\alpha\pm\beta)=\cos\alpha\cos\beta\mp\sin\alpha\sin\beta$ |
| Ángulo doble | $\sin2\theta=2\sin\theta\cos\theta$, $\;\cos2\theta=\cos^2\theta-\sin^2\theta$ |
| Cuadrados | $\sin^2\theta=\dfrac{1-\cos2\theta}{2}$, $\;\cos^2\theta=\dfrac{1+\cos2\theta}{2}$ |
| Producto a suma | $\sin\alpha\sin\beta=\tfrac12[\cos(\alpha-\beta)-\cos(\alpha+\beta)]$ |
| Producto a suma | $\cos\alpha\cos\beta=\tfrac12[\cos(\alpha-\beta)+\cos(\alpha+\beta)]$ |
| Producto a suma | $\sin\alpha\cos\beta=\tfrac12[\sin(\alpha+\beta)+\sin(\alpha-\beta)]$ |
| Forma amplitud-fase | $a\cos\theta+b\sin\theta=A\cos(\theta-\varphi)$, con $A=\sqrt{a^2+b^2}$, $\varphi=\arctan\dfrac ba$ (si $a>0$) |
| Múltiplos de $\pi$ | $\sin(n\pi)=0$, $\;\cos(n\pi)=(-1)^n$ ($n$ entero) |

La forma amplitud-fase es la tercera representación de la serie de Fourier: cada armónico se escribe como un solo coseno $A_n\cos(n\omega_0t-\varphi_n)$ con $A_n=\sqrt{a_n^2+b_n^2}=2\lvert C_n\rvert$. En este ejercicio $A_n=\lvert b_n\rvert=2/n$.

### A.4. Paridad e integrales sobre un intervalo simétrico

| Identidad | Expresión |
|---|---|
| Función par | $g(-t)=g(t)\;\Rightarrow\;\displaystyle\int_{-a}^{a}g\,dt=2\int_0^ag\,dt$ |
| Función impar | $g(-t)=-g(t)\;\Rightarrow\;\displaystyle\int_{-a}^{a}g\,dt=0$ |
| Producto | par·par = par; impar·impar = par; par·impar = impar |
| Serie de una función par | $b_n=0$ (solo cosenos y término DC) |
| Serie de una función impar | $a_n=0$ (solo senos) |
| Simetría de media onda | si $f(t+T/2)=-f(t)$, solo hay armónicos impares |

### A.5. Ortogonalidad de los armónicos

Estas integrales son las que justifican las fórmulas de $a_n$, $b_n$ y $C_n$. Para $m,n$ enteros positivos y un período completo:

$$
\frac{2}{T}\int_{-T/2}^{T/2}\cos(m\omega_0t)\cos(n\omega_0t)\,dt=\delta_{mn}
\qquad
\frac{2}{T}\int_{-T/2}^{T/2}\sin(m\omega_0t)\sin(n\omega_0t)\,dt=\delta_{mn}
$$

$$
\int_{-T/2}^{T/2}\sin(m\omega_0t)\cos(n\omega_0t)\,dt=0
\qquad
\frac1T\int_{-T/2}^{T/2}e^{jm\omega_0t}\,e^{-jn\omega_0t}\,dt=\delta_{mn}
$$

donde $\delta_{mn}=1$ si $m=n$ y $0$ si $m\neq n$. **Idea:** multiplicar la serie por $e^{-jn\omega_0t}$ e integrar "filtra" el único término $n$ y anula todos los demás; por eso $C_n$ se obtiene con esa integral.

### A.6. Integrales usadas en el cálculo

| Identidad | Expresión |
|---|---|
| Por partes | $\displaystyle\int u\,dv=uv-\int v\,du$ |
| Seno | $\displaystyle\int\sin(nt)\,dt=-\dfrac{\cos(nt)}{n}$ |
| Coseno | $\displaystyle\int\cos(nt)\,dt=\dfrac{\sin(nt)}{n}$ |
| Exponencial | $\displaystyle\int e^{at}\,dt=\dfrac{e^{at}}{a}$ (con $a=-jn$ en este ejercicio) |
| $t$ por seno | $\displaystyle\int t\sin(nt)\,dt=-\dfrac{t\cos(nt)}{n}+\dfrac{\sin(nt)}{n^2}$ |
| $t$ por coseno | $\displaystyle\int t\cos(nt)\,dt=\dfrac{t\sin(nt)}{n}+\dfrac{\cos(nt)}{n^2}$ |
| $t$ por exponencial | $\displaystyle\int te^{at}\,dt=\dfrac{te^{at}}{a}-\dfrac{e^{at}}{a^2}$ |

La última fila resume 4.1: con $a=-jn$ se tiene $a^2=-n^2$, y queda $-\dfrac{te^{-jnt}}{jn}+\dfrac{e^{-jnt}}{n^2}$.

### A.7. Parseval: potencia de la señal

La potencia media de una señal periódica se calcula igual en el tiempo y en los coeficientes:

$$
\boxed{\frac1T\int_{-T/2}^{T/2}\lvert f(t)\rvert^2\,dt=\sum_{n=-\infty}^{\infty}\lvert C_n\rvert^2=\frac{a_0^2}{4}+\frac12\sum_{n=1}^{\infty}\left(a_n^2+b_n^2\right)}
$$

Para este ejercicio, por el lado del tiempo:

$$
\frac{1}{2\pi}\int_{-\pi}^{\pi}t^2\,dt=\frac1{2\pi}\cdot\frac{2\pi^3}{3}=\frac{\pi^2}{3}
$$

y por el lado de los coeficientes, con $\lvert C_n\rvert=1/\lvert n\rvert$ y $\sum_{n=1}^\infty 1/n^2=\pi^2/6$:

$$
\sum_{n\neq0}\frac1{n^2}=2\sum_{n=1}^{\infty}\frac1{n^2}=2\cdot\frac{\pi^2}{6}=\frac{\pi^2}{3}\;\checkmark
$$

Coinciden, lo que confirma los coeficientes por un camino independiente. Además, la potencia captada por los primeros $N$ armónicos es $2\sum_{n=1}^N 1/n^2$, útil para decidir cuántos armónicos conservar.

```python
# Verificación numérica de ortogonalidad y de Parseval (sin bucles)
tt = np.linspace(-np.pi, np.pi, 200_000, endpoint=False)
dt = tt[1] - tt[0]

# Ortogonalidad: (1/pi) * integral de sin(m t) sin(n t) para m, n = 1..3
m = np.arange(1, 4).reshape(-1, 1, 1)
n = np.arange(1, 4).reshape(1, -1, 1)
G = np.sum(np.sin(m * tt) * np.sin(n * tt), axis=2) * dt / np.pi
print("Matriz de ortogonalidad (debe ser la identidad):")
print(np.round(G, 4))

# Parseval: potencia en el tiempo vs. en los coeficientes
potencia_tiempo = np.sum(tt ** 2) * dt / (2 * np.pi)
k = np.arange(1, 100_001)
acumulada = 2 * np.cumsum(1 / k ** 2)       # potencia captada por los primeros N armónicos
print()
print(f"Potencia en el tiempo   : {potencia_tiempo:.5f}")
print(f"Potencia en coeficientes: {acumulada[-1]:.5f}")
print(f"pi^2 / 3 (teórico)      : {np.pi ** 2 / 3:.5f}")

# Porcentaje de la potencia total captado con N armónicos
N_lista = np.array([1, 3, 5, 10, 50])
captada = acumulada[N_lista - 1] / (np.pi ** 2 / 3) * 100
print()
print("N armónicos      :", N_lista)
print("Potencia captada :", np.round(captada, 2), "%")
```
