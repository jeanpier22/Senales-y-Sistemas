### Señal continua y discreta

Una señal continua está definida para cualquier instante de tiempo \(t\):

$$
\boxed{x(t)=\sin(2\pi ft)}
$$

Una señal discreta se obtiene al tomar muestras de la señal en instantes separados por un período de muestreo \(T_s\):

$$
\boxed{x[n]=x(nT_s)}
$$

Para este ejemplo:

$$
f=5\,Hz
\qquad
T_s=0.05\,s
$$

Por lo tanto:

$$
\boxed{x[n]=\sin(2\pi f nT_s)}
$$

El siguiente código permite visualizar la señal continua y su representación discreta.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ==============================
# Señal continua
# ==============================
t = np.linspace(0, 1, 1000)

f = 5  # Frecuencia de 5 Hz
x_cont = np.sin(2 * np.pi * f * t)

# ==============================
# Señal discreta
# ==============================
n = np.arange(0, 21)

Ts = 0.05  # Periodo de muestreo
x_disc = np.sin(2 * np.pi * f * n * Ts)

# ==============================
# Gráficas
# ==============================
plt.figure(figsize=(10, 6))

# Señal continua
plt.subplot(2, 1, 1)
plt.plot(t, x_cont)
plt.title("Señal continua")
plt.xlabel("Tiempo t [s]")
plt.ylabel("Amplitud")
plt.grid(True)

# Señal discreta
plt.subplot(2, 1, 2)
plt.stem(n * Ts, x_disc)
plt.title("Señal discreta")
plt.xlabel("Tiempo nTs [s]")
plt.ylabel("Amplitud")
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Señal real y señal compleja

Una señal real únicamente presenta valores reales y puede representarse como:

$$
\boxed{x(t)\in\mathbb{R}}
$$

Una señal compleja puede expresarse como:

$$
\boxed{x(t)=x_R(t)+jx_I(t)}
$$

donde \(x_R(t)\) es la **parte real** y \(x_I(t)\) es la **parte imaginaria**.

Para este ejemplo se utiliza la señal compleja:

$$
\boxed{x(t)=e^{j2\pi ft}}
$$

Aplicando la identidad de Euler:

$$
\boxed{x(t)=\cos(2\pi ft)+j\sin(2\pi ft)}
$$

El siguiente código permite comparar una señal real con las partes real e imaginaria de una señal compleja.


```python
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# Eje temporal
# ==============================
t = np.linspace(0, 1, 1000)

# ==============================
# Señal real
# ==============================
f = 5
x_real = np.sin(2 * np.pi * f * t)

# ==============================
# Señal compleja
# ==============================
fc = 5

x_compleja = np.exp(1j * 2 * np.pi * fc * t)

# Separar parte real e imaginaria
parte_real = np.real(x_compleja)
parte_imaginaria = np.imag(x_compleja)

# ==============================
# Gráficas
# ==============================
plt.figure(figsize=(10, 8))

# Señal real
plt.subplot(3, 1, 1)
plt.plot(t, x_real)
plt.title("Señal real")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")
plt.grid(True)

# Parte real de la señal compleja
plt.subplot(3, 1, 2)
plt.plot(t, parte_real)
plt.title("Parte real de la señal compleja")
plt.xlabel("Tiempo [s]")
plt.ylabel("Re{x(t)}")
plt.grid(True)

# Parte imaginaria de la señal compleja
plt.subplot(3, 1, 3)
plt.plot(t, parte_imaginaria)
plt.title("Parte imaginaria de la señal compleja")
plt.xlabel("Tiempo [s]")
plt.ylabel("Im{x(t)}")
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Representación de una señal compleja en el plano complejo

Una señal compleja puede representarse mediante sus componentes real e imaginaria:

$$
\boxed{z(t)=x(t)+jy(t)}
$$

donde:

$$
x(t)=\operatorname{Re}\{z(t)\}
$$

$$
y(t)=\operatorname{Im}\{z(t)\}
$$

Para la señal:

$$
\boxed{z(t)=e^{j2\pi ft}}
$$

cada muestra puede representarse como un punto en el plano complejo, donde el eje horizontal corresponde a la **parte real** y el eje vertical a la **parte imaginaria**.

El siguiente código representa las muestras de la señal compleja como puntos en el plano cartesiano.


```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
f = 1
t = np.linspace(0, 1, 21)

# Señal compleja
z = np.exp(1j * 2 * np.pi * f * t)

# Parte real e imaginaria
x = np.real(z)
y = np.imag(z)

# Plano cartesiano
plt.figure(figsize=(7, 7))

plt.scatter(x, y, s=50)

# Ejes
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlabel("Parte real")
plt.ylabel("Parte imaginaria")
plt.title("Señal compleja representada por puntos")

plt.axis("equal")
plt.grid(True)

plt.show()
```

### Señal determinista y aleatoria

Una **señal determinista** es aquella cuyo valor puede determinarse exactamente en cualquier instante de tiempo mediante una expresión matemática o una regla conocida.

Por ejemplo:

$$
\boxed{x(t)=\sin(2\pi ft)}
$$

Una **señal aleatoria** presenta valores que no pueden predecirse exactamente para cada instante de tiempo y se describe mediante conceptos probabilísticos.

En el siguiente ejemplo se compara una señal sinusoidal determinista con una señal generada aleatoriamente mediante una distribución normal.


```python
import numpy as np
import matplotlib.pyplot as plt

# Eje temporal
t = np.linspace(0, 1, 500)

# ==============================
# Señal determinista
# ==============================
f = 5
x_determinista = np.sin(2 * np.pi * f * t)

# ==============================
# Señal aleatoria
# ==============================
np.random.seed(10)  # Para obtener siempre el mismo resultado
x_aleatoria = np.random.randn(len(t))

# ==============================
# Gráficas
# ==============================
plt.figure(figsize=(10, 6))

# Determinista
plt.subplot(2, 1, 1)
plt.plot(t, x_determinista)
plt.title("Señal determinista")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")
plt.grid(True)

# Aleatoria
plt.subplot(2, 1, 2)
plt.plot(t, x_aleatoria)
plt.title("Señal aleatoria")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud")
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Señal par

Una señal es **par** cuando presenta simetría respecto al eje vertical y cumple:

$$
\boxed{x(t)=x(-t)}
$$

En este ejemplo se utiliza la función:

$$
\boxed{x(t)=\cos(t)}
$$

Como:

$$
\cos(t)=\cos(-t)
$$

se cumple la condición de simetría par.

El siguiente código permite visualizar una señal par y comprobar gráficamente su simetría respecto al eje \(t=0\).


```python
import numpy as np
import matplotlib.pyplot as plt

# Tiempo
t = np.linspace(-2*np.pi, 2*np.pi, 500)

# Señal par
x_par = np.cos(t)

# Graficar
plt.figure(figsize=(9, 4))

plt.plot(t, x_par)

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlabel("t")
plt.ylabel("x(t)")
plt.title("Señal Par: x(t) = x(-t)")
plt.grid(True)

plt.show()
```

### Señal impar

Una señal es **impar** cuando presenta simetría respecto al origen y cumple:

$$
\boxed{x(t)=-x(-t)}
$$

En este ejemplo se utiliza la función:

$$
\boxed{x(t)=\sin(t)}
$$

Como:

$$
\sin(t)=-\sin(-t)
$$

se cumple la condición de simetría impar.

El siguiente código permite visualizar una señal impar y observar su simetría respecto al origen.


```python
import numpy as np
import matplotlib.pyplot as plt

# Tiempo
t = np.linspace(-2*np.pi, 2*np.pi, 500)

# Señal impar
x_impar = np.sin(t)

# Graficar
plt.figure(figsize=(9, 4))

plt.plot(t, x_impar)

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlabel("t")
plt.ylabel("x(t)")
plt.title("Señal Impar: x(t) = -x(-t)")
plt.grid(True)

plt.show()
```

### Descomposición en parte par e impar

Una señal puede ser descompuesta en una **parte par** y una **parte impar**:

$$
\boxed{x(t)=x_p(t)+x_i(t)}
$$

donde:

$$
\boxed{x_p(t)=\frac{x(t)+x(-t)}{2}}
$$

y:

$$
\boxed{x_i(t)=\frac{x(t)-x(-t)}{2}}
$$

Para este ejemplo se considera la señal:

$$
\boxed{x(t)=t^2+t}
$$

Al evaluar la señal en \(-t\):

$$
x(-t)=t^2-t
$$

Por lo tanto, su parte par es:

$$
\boxed{x_p(t)=t^2}
$$

y su parte impar:

$$
\boxed{x_i(t)=t}
$$

De esta manera se verifica que:

$$
\boxed{x(t)=t^2+t=x_p(t)+x_i(t)}
$$

El siguiente código permite visualizar la señal original y su descomposición en parte par e impar.


```python
import numpy as np
import matplotlib.pyplot as plt

# ==============================
# Eje temporal
# ==============================
t = np.linspace(-5, 5, 500)

# ==============================
# Señal original
# ==============================
x = t**2 + t

# Señal evaluada en -t
x_neg = (-t)**2 + (-t)

# ==============================
# Parte par
# ==============================
x_par = (x + x_neg) / 2

# ==============================
# Parte impar
# ==============================
x_impar = (x - x_neg) / 2

# ==============================
# Gráficas
# ==============================
plt.figure(figsize=(10, 7))

# Señal original
plt.subplot(3, 1, 1)
plt.plot(t, x)
plt.title("Señal original: x(t) = t² + t")
plt.ylabel("x(t)")
plt.grid(True)

# Parte par
plt.subplot(3, 1, 2)
plt.plot(t, x_par)
plt.title("Parte par: xp(t) = t²")
plt.ylabel("xp(t)")
plt.grid(True)

# Parte impar
plt.subplot(3, 1, 3)
plt.plot(t, x_impar)
plt.title("Parte impar: xi(t) = t")
plt.xlabel("t")
plt.ylabel("xi(t)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Reconstrucción de la señal
x_reconstruida = x_par + x_impar

# Verificación
print("¿La señal original es igual a la reconstruida?")
print(np.allclose(x, x_reconstruida))
```

### Simetría conjugada

Una señal compleja presenta **simetría conjugada** cuando cumple:

$$
\boxed{x(t)=x^*(-t)}
$$

donde \(x^*(t)\) representa el **conjugado complejo** de la señal.

Para este ejemplo se utiliza:

$$
\boxed{x(t)=e^{j2\pi ft}}
$$

Aplicando la identidad de Euler:

$$
x(t)=\cos(2\pi ft)+j\sin(2\pi ft)
$$

Al evaluar en \(-t\):

$$
x(-t)=e^{-j2\pi ft}
$$

y tomando su conjugado:

$$
\boxed{x^*(-t)=e^{j2\pi ft}=x(t)}
$$

Por lo tanto, se cumple la condición de **simetría conjugada**.

El siguiente código representa mediante puntos la señal \(x(t)\) y su conjugado \(x^*(-t)\) en el plano complejo.


```python
import numpy as np
import matplotlib.pyplot as plt

# Tiempo
t = np.linspace(-1, 1, 21)

# Señal compleja
f = 1
x = np.exp(1j * 2 * np.pi * f * t)

# Conjugado de x(-t)
x_conj_neg = np.conj(np.exp(1j * 2 * np.pi * f * (-t)))

# Graficar señal
plt.figure(figsize=(7, 7))

plt.scatter(np.real(x), np.imag(x), label='x(t)')

# Graficar x*(-t)
plt.scatter(
    np.real(x_conj_neg),
    np.imag(x_conj_neg),
    marker='x',
    label='x*(-t)'
)

# Ejes
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)

plt.xlabel('Parte real')
plt.ylabel('Parte imaginaria')
plt.title('Simetría conjugada')
plt.axis('equal')
plt.grid(True)
plt.legend()

plt.show()
```
