## 4.-  Funciones, graficar, carga de datos
Las funciones son una parte importante de cualquier lenguaje que las admita. Nos permiten encapsular y modularizar nuestro código para su reutilización flexible y repetida. En lugar de reescribir o copiar el mismo código para realizar una tarea común para nosotros, simplemente podemos llamar a nuestra función. A continuación, definimos dos funciones, seno y coseno, que devuelven los valores de seno y coseno dada una matriz de valores de entrada y una frecuencia analógica lineal deseada (Hz).

$$f(t) = sin(2\pi t f)$$
$$f(t) = cos(2\pi t f)$$

```
def sine(t,f):
  x = np.sin(2*np.pi*f*t)
  return x

def cosine(t,f):
    return np.cos(2*np.pi*f*t)
```

```
t = np.linspace(0,5,1000) #1000 evenly spaced points from 0 to 5
frequency = 2 #2 Hz frequency

s = sine(t,frequency)
c = cosine(t,frequency)
```

Calcular $y(t)$
$$x_1(t)  = sin(2\pi f_1 t) $$
$$x_2(t)  = sin(2\pi f_2 t) $$
$$x_3(t)  = sin(2\pi f_3 t) $$
$$y(t) = x_1(t) + x_2(t) + x_3(t) $$

para $t\in  [0; 5]$
* $f_1  = 2 $, $f_2  = 3 $, $f_3  = 4 $
* $f_1  = 4 $, $f_2  = 6 $, $f_3  = 8 $
* $f_1  = 10 $, $f_2  = 20 $, $f_3  = 30 $


```
def seno(f,t,a):
  x1 = a*np.sin(2*np.pi*f*t)
  return x1
```

```
def ejercicio1(f1,f2,f3,t):
  x1 = seno(f1,t,2)
  x2 = seno(f2,t,3)
  x3 = seno(f3,t,4)
  y  = x1 + x2 + x3
  return y
```

# Crear una función que retorne la suma de una señal seno con un coseno al cuadrado, ambas de 30 Hz. Genere su gráfica.

```
def suma_seno_coseno(f1, f2, t):
  x1 = np.sin(2*np.pi*f1*t)
  x2 = np.cos(2*np.pi*f2*t)**2
  return x1+x2

t = np.linspace(0,0.5,100)
y = suma_seno_coseno(30,30,t)

plt.figure(figsize=(10, 5))
plt.plot(t, y, 'r', label='suma$', linewidth=1)  # Added linewidth
plt.title('Suma') #title
plt.xlabel('Time (s)') #time
plt.ylabel('Signal Value') #ylabel
plt.axis([0, 0.2, -1, 1.5])
plt.legend() #legend that uses the "label" identifiers from above

```

# 2. Función Rampa

```
import numpy as np
import matplotlib.pyplot as plt

# Crear un vector de tiempo de -5 a 5 segundos con paso de 0.1 segundos
t = np.arange(-7, 5, 0.1)

# Calcular la respuesta de la función de rampa con pendiente de 2
pendiente = 1/3
rampa = np.maximum(-1, pendiente * (t+3))

# Graficar la función de rampa con pendiente de 2
plt.plot(t, rampa, label='Rampa con pendiente de 2')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.title('Función de Rampa con Pendiente de 2')
plt.grid(True)

```

# 3. Función Rampa con pendiente negativa

```
import numpy as np
import matplotlib.pyplot as plt
# Parámetros de la recta
pendiente = -1
interseccion = 1  # Punto de intersección con el eje t

# Crear un vector de tiempo de -5 a 5 segundos con paso de 0.1 segundos
t = np.arange(-5, 5, 0.1)

# Calcular la respuesta de la rampa
rampa_resultado = np.maximum(0, pendiente * (t - interseccion)*(t>-1))

# Graficar la función de la rampa
plt.plot(t, rampa_resultado, label='Rampa: y = {}x + {}'.format(pendiente, interseccion))
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.title('Rampa con Pendiente Negativa')
plt.grid(True)
plt.legend()
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de la recta
pendiente = 2
interseccion = 1  # Punto de intersección con el eje t

# Crear un vector de tiempo de -5 a 5 segundos con paso de 0.1 segundos
t = np.arange(-5, 5, 0.1)

# Calcular la respuesta de la recta y de la rampa
recta_resultado = pendiente * (t - interseccion)
rampa_resultado = np.maximum(0, pendiente * (t - interseccion)) * (t > interseccion)

# Graficar la función de la recta y de la rampa
plt.plot(t, recta_resultado, label='Recta: y = {}x + {}'.format(pendiente, interseccion))
plt.plot(t, rampa_resultado, label='Rampa: t > 1, y = {}x + {}'.format(pendiente, interseccion))
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.title('Recta y Rampa')
plt.grid(True)
plt.legend()
plt.show()
```

# 4. Pulso Rectangular

```
from typing_extensions import Type
from typing import TypeVar
def pulso_cuadrado(t, t0=0., t1 = 1.):
  u0 = (t>=t0)
  print(type(u0))
  print((u0.dtype))
  u0 = (t>=t0) +0.
  print(u0.dtype)
  u1 = (t>=t1) +0.
  s = u0 - u1
  return s
```

```
t = np.arange(-5, 5, 0.1)
t0=-2
t1=0
s = pulso_cuadrado(t,t0,t1)

plt.figure()
plt.plot(t, s, color = 'red', label= r'$s_0(t) = \Pi (\frac{t-1}{2})$')
plt.axis([-4,4,-1,2])
plt.xlabel('Time(s)')
plt.ylabel('Amplitud(V)')
plt.title('Pulso Rectangular')
plt.legend()
```

Otra función de trazado que encontraremos particularmente útil es $\textrm{plt.stem ()} $. Esta función proporciona líneas discretas para cada punto de nuestra señal. Tenga cuidado al usar stem cuando tenga señales largas, ya que la grafica puede estar desordenada. Por lo tanto, para señales más largas (mayor que la longitud 30, aproximadamente) $\textrm {plt.plot ()} $ es lo mejor.

```
toy_signal = [1,4,-3,8,-4,7,9]
x_vals = range(len(toy_signal)) #x-axis values
plt.figure()
plt.title('Gross Plot')
plt.plot(x_vals,toy_signal) #this plot looks unnatural
plt.xlabel('Index')
plt.ylabel('Some units')

#now let's try stem
plt.figure()
plt.title('Prettier Plot')
plt.stem(x_vals,toy_signal) #pretty!
plt.xlabel('Index')
plt.ylabel('Some units')
```
