## 2. Señal continua y señal discreta

Una **señal continua** está definida para cualquier instante de tiempo $t$:

$$
\boxed{x(t) = \sin(2\pi f t)}
$$

Una **señal discreta** se obtiene al tomar muestras de la señal en instantes separados
por un período de muestreo $T_s$:

$$
\boxed{x[n] = x(n\,T_s)}
$$

Nota la diferencia de notación: el paréntesis $x(t)$ indica variable continua; el corchete
$x[n]$ indica que $n$ es un **entero**, un número de muestra.

Para este ejemplo:

$$
f = 5\ \text{Hz}, \qquad T_s = 0{,}05\ \text{s}
$$

de donde:

$$
\boxed{x[n] = \sin(2\pi f\,n\,T_s)}
$$

Con $T_s = 0{,}05$ s la frecuencia de muestreo es $f_m = 1/T_s = 20$ Hz, es decir
**4 muestras por período**.

```python
f = 5.0        # frecuencia de la señal, Hz
Ts = 0.05      # período de muestreo, s
fm = 1 / Ts    # frecuencia de muestreo, Hz

# --- Señal continua ---------------------------------------------------
t = np.linspace(0, 1, 1000)
x_continua = np.sin(2 * np.pi * f * t)

# --- Señal discreta ---------------------------------------------------
n = np.arange(0, 21)                      # índices de muestra: 0, 1, 2, ... 20
x_discreta = np.sin(2 * np.pi * f * n * Ts)

print("Frecuencia de muestreo fm =", fm, "Hz")
print("Muestras por período      =", fm / f)
print("Número total de muestras  =", len(n))

plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.plot(t, x_continua)
plt.title(f"Señal continua x(t), f = {f} Hz")
plt.ylabel("Amplitud [V]")

plt.subplot(2, 1, 2)
plt.stem(n * Ts, x_discreta)
plt.title(f"Señal discreta x[n] = x(n Ts), con Ts = {Ts} s")
plt.xlabel("Tiempo n Ts [s]")
plt.ylabel("Amplitud [V]")

plt.tight_layout()
plt.show()
```

## 3. Señal real y señal compleja

Una **señal real** solo toma valores reales:

$$
\boxed{x(t) \in \mathbb{R}}
$$

Una **señal compleja** tiene una parte real y una imaginaria:

$$
\boxed{x(t) = x_R(t) + j\,x_I(t)}
$$

La señal compleja fundamental del curso es la **exponencial compleja**:

$$
\boxed{x(t) = e^{j2\pi f t}}
$$

que por la identidad de Euler se escribe:

$$
\boxed{x(t) = \cos(2\pi f t) + j\,\sin(2\pi f t)}
$$

Es decir: **una sola** exponencial compleja lleva a la vez el coseno y el seno. Esa es la
razón de que casi todo el análisis de señales se haga con exponenciales complejas en lugar
de con senos y cosenos por separado.

En Python la unidad imaginaria se escribe `1j`, y las partes se extraen con `np.real()`
y `np.imag()`.

```python
f = 5.0
t = np.linspace(0, 1, 1000)

# --- Señal real -------------------------------------------------------
x_real = np.sin(2 * np.pi * f * t)

# --- Señal compleja ---------------------------------------------------
x_compleja = np.exp(1j * 2 * np.pi * f * t)

parte_real = np.real(x_compleja)
parte_imaginaria = np.imag(x_compleja)

print("Tipo de x_real    :", x_real.dtype)
print("Tipo de x_compleja:", x_compleja.dtype)

plt.figure(figsize=(10, 8))

plt.subplot(3, 1, 1)
plt.plot(t, x_real)
plt.title("Señal real x(t) = sen(2 pi f t)")
plt.ylabel("Amplitud [V]")

plt.subplot(3, 1, 2)
plt.plot(t, parte_real)
plt.title("Parte real de la exponencial compleja: es el coseno")
plt.ylabel("Re{x(t)}")

plt.subplot(3, 1, 3)
plt.plot(t, parte_imaginaria)
plt.title("Parte imaginaria de la exponencial compleja: es el seno")
plt.xlabel("Tiempo [s]")
plt.ylabel("Im{x(t)}")

plt.tight_layout()
plt.show()
```

### 3.1. Representación en el plano complejo

Una señal compleja puede representarse mediante sus componentes real e imaginaria:

$$
\boxed{z(t) = x(t) + j\,y(t)}
$$

donde $x(t) = \operatorname{Re}\{z(t)\}$ e $y(t) = \operatorname{Im}\{z(t)\}$.

Para $z(t) = e^{j2\pi f t}$, cada muestra es un punto del plano complejo: el eje horizontal
es la parte real y el vertical la imaginaria.

Como $|e^{j\theta}| = 1$ para todo $\theta$, **todos los puntos caen sobre la circunferencia
de radio 1**. Al avanzar el tiempo, el punto recorre esa circunferencia en sentido
antihorario, dando una vuelta completa cada período $T = 1/f$.

```python
f = 1.0
t = np.linspace(0, 1, 21)     # un período completo, 21 muestras

z = np.exp(1j * 2 * np.pi * f * t)

# Circunferencia de radio 1, solo como referencia
angulo = np.linspace(0, 2 * np.pi, 400)

plt.figure(figsize=(6.5, 6.5))

plt.plot(np.cos(angulo), np.sin(angulo), "--", color="gray",
         label="Circunferencia de radio 1")
plt.scatter(np.real(z), np.imag(z), s=60, label="Muestras de z(t)")

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.title("Muestras de la exponencial compleja en el plano complejo")
plt.xlabel("Parte real")
plt.ylabel("Parte imaginaria")
plt.axis("equal")
plt.legend(loc="upper right")
plt.show()

print("Módulo de la primera muestra:", np.abs(z[0]))
print("Módulo de la última muestra :", np.abs(z[-1]))
```

## 4. Señal determinista y señal aleatoria

Una **señal determinista** puede calcularse exactamente en cualquier instante mediante una
expresión matemática conocida. Por ejemplo:

$$
\boxed{x(t) = \sin(2\pi f t)}
$$

Si se evalúa dos veces, da lo mismo las dos veces.

Una **señal aleatoria** toma valores que no pueden predecirse instante a instante, y solo
se describe mediante conceptos probabilísticos: media, varianza, distribución.

En el código, `np.random.seed(10)` fija la semilla del generador para que el ejemplo salga
igual en todas las computadoras. Cuidado con la idea: la señal sigue siendo aleatoria; lo
que se repite es la secuencia concreta que genera esta ejecución.

```python
t = np.linspace(0, 1, 500)

# --- Determinista -----------------------------------------------------
f = 5.0
x_determinista = np.sin(2 * np.pi * f * t)

# --- Aleatoria --------------------------------------------------------
np.random.seed(10)                       # solo para reproducir el mismo dibujo
x_aleatoria = np.random.randn(500)       # distribución normal, media 0, varianza 1

print("Determinista -> media =", x_determinista.mean())
print("Aleatoria    -> media =", x_aleatoria.mean())

plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.plot(t, x_determinista)
plt.title(f"Señal determinista: senoide de {f} Hz")
plt.ylabel("Amplitud [V]")

plt.subplot(2, 1, 2)
plt.plot(t, x_aleatoria)
plt.title("Señal aleatoria: ruido gaussiano de media 0")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")

plt.tight_layout()
plt.show()
```

## 5. Señales pares e impares

### 5.1. Señal par

Una señal es **par** cuando es simétrica respecto al eje vertical:

$$
\boxed{x(t) = x(-t)}
$$

El ejemplo canónico es el coseno, porque $\cos(t) = \cos(-t)$.

### 5.2. Señal impar

Una señal es **impar** cuando es simétrica respecto al origen:

$$
\boxed{x(t) = -x(-t)}
$$

El ejemplo canónico es el seno, porque $\sin(-t) = -\sin(t)$.

Una consecuencia inmediata: toda señal impar continua en el origen cumple $x(0) = 0$.

En las gráficas siguientes se dibuja cada señal junto con su reflejada, para comprobar
que efectivamente coinciden.

```python
t = np.linspace(-2 * np.pi, 2 * np.pi, 500)

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(t, np.cos(t), linewidth=3, label="x(t) = cos(t)")
plt.plot(t, np.cos(-t), "--", label="x(-t)")
plt.title("Señal par: x(t) = x(-t)")
plt.xlabel("t [s]")
plt.ylabel("Amplitud")
plt.axvline(0, color="black", linewidth=0.8)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(t, np.sin(t), linewidth=3, label="x(t) = sen(t)")
plt.plot(t, -np.sin(-t), "--", label="-x(-t)")
plt.title("Señal impar: x(t) = -x(-t)")
plt.xlabel("t [s]")
plt.ylabel("Amplitud")
plt.axvline(0, color="black", linewidth=0.8)
plt.legend()

plt.tight_layout()
plt.show()
```

### 5.3. Descomposición en parte par e impar

**Toda** señal —sea o no simétrica— puede escribirse como la suma de una parte par y una
parte impar:

$$
\boxed{x(t) = x_p(t) + x_i(t)}
$$

donde:

$$
\boxed{x_p(t) = \frac{x(t) + x(-t)}{2}}
\qquad\qquad
\boxed{x_i(t) = \frac{x(t) - x(-t)}{2}}
$$

Para verlo, basta sumar las dos expresiones: los términos $x(-t)$ se cancelan y queda
$x_p(t) + x_i(t) = x(t)$.

Como ejemplo tomamos:

$$
\boxed{x(t) = t^2 + t}
$$

Al evaluar en $-t$:

$$
x(-t) = (-t)^2 + (-t) = t^2 - t
$$

Por lo tanto:

$$
x_p(t) = \frac{(t^2+t) + (t^2-t)}{2} = \boxed{t^2}
\qquad\qquad
x_i(t) = \frac{(t^2+t) - (t^2-t)}{2} = \boxed{t}
$$

Lo cual tiene sentido: $t^2$ es par y $t$ es impar.

```python
def senal(t):
    """Señal de ejemplo: x(t) = t^2 + t."""
    x = t**2 + t
    return x


t = np.linspace(-5, 5, 500)

x = senal(t)
x_reflejada = senal(-t)          # esto es x(-t)

x_par = (x + x_reflejada) / 2
x_impar = (x - x_reflejada) / 2

plt.figure(figsize=(10, 9))

plt.subplot(3, 1, 1)
plt.plot(t, x)
plt.title("Señal original x(t) = t^2 + t")
plt.ylabel("x(t)")

plt.subplot(3, 1, 2)
plt.plot(t, x_par)
plt.title("Parte par xp(t) = t^2")
plt.ylabel("xp(t)")

plt.subplot(3, 1, 3)
plt.plot(t, x_impar)
plt.title("Parte impar xi(t) = t")
plt.xlabel("t [s]")
plt.ylabel("xi(t)")

plt.tight_layout()
plt.show()
```

```python
# Verificación numérica de la descomposición
x_reconstruida = x_par + x_impar

print("¿x(t) = xp(t) + xi(t)?      ", np.allclose(x, x_reconstruida))
print("¿La parte par vale t^2?     ", np.allclose(x_par, t**2))
print("¿La parte impar vale t?     ", np.allclose(x_impar, t))
```

## 6. Simetría conjugada

Una señal compleja presenta **simetría conjugada** (también llamada *hermítica*) cuando:

$$
\boxed{x(t) = x^*(-t)}
$$

donde $x^*(t)$ es el **conjugado complejo**: se le cambia el signo a la parte imaginaria.

Para la exponencial compleja $x(t) = e^{j2\pi f t}$:

$$
x(-t) = e^{-j2\pi f t}
\qquad\Longrightarrow\qquad
x^*(-t) = e^{j2\pi f t} = x(t)
$$

de modo que **sí** cumple la condición.

En la gráfica, los puntos de $x(t)$ y los de $x^*(-t)$ caen exactamente unos sobre otros:
por eso se dibujan los segundos con una marca más pequeña, para que se vea la coincidencia
en vez de que una serie tape a la otra.

```python
f = 1.0
t = np.linspace(-1, 1, 21)

x = np.exp(1j * 2 * np.pi * f * t)
x_conjugada_reflejada = np.conj(np.exp(1j * 2 * np.pi * f * (-t)))

print("¿Se cumple x(t) = x*(-t)?", np.allclose(x, x_conjugada_reflejada))

plt.figure(figsize=(6.5, 6.5))

plt.scatter(np.real(x), np.imag(x), s=150, label="x(t)")
plt.scatter(np.real(x_conjugada_reflejada), np.imag(x_conjugada_reflejada),
            s=30, marker="x", color="red", label="x*(-t)")

plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.title("Simetría conjugada: las dos series coinciden punto a punto")
plt.xlabel("Parte real")
plt.ylabel("Parte imaginaria")
plt.axis("equal")
plt.legend()
plt.show()
```

### 7.2. Cargar el audio

Ajusta `RUTA_AUDIO` a la ubicación real del archivo en tu Drive.

La frecuencia de muestreo del archivo se lee del propio audio gracias a `sr=None`. Sin ese
argumento, `librosa` remuestrearía a 22 050 Hz por defecto y la frecuencia declarada no
sería la real.

A partir de ella, la duración del registro es:

$$

$$

```python
import librosa

RUTA_AUDIO = "/content/drive/MyDrive/Cursos/Señales y Sistemas 2026 - 2/3 segundos con voz.mp3"

amplitudes, fm = librosa.load(RUTA_AUDIO, sr=None)   # sr=None conserva la fm original

N = len(amplitudes)
tf = N / fm

print("Frecuencia de muestreo  :", fm, "Hz")
print("Número total de muestras:", N)
print("Duración del registro   :", tf, "s")
```

> **Si esta celda da `FileNotFoundError`**, el archivo no está en esa ruta. Comprueba en el

> y usar solo su nombre como ruta.

```python
t = np.arange(N) / fm          # instante de cada muestra, en segundos

plt.figure(figsize=(14, 4))
plt.plot(t, amplitudes, linewidth=0.5)
plt.title(f"Forma de onda del audio, fm = {fm} Hz")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud normalizada")
plt.show()
```

### 7.3. La señal es discreta

En la vista de tres segundos completos no se nota, porque las muestras quedan solapadas.
Al ampliar 20 ms sí se distinguen una por una: un archivo de audio **es** una señal discreta.

```python
inicio = 1.0        # segundos
duracion = 0.02     # segundos

primera = int(inicio * fm)
ultima = int((inicio + duracion) * fm)

plt.figure(figsize=(12, 4))
plt.plot(t[primera:ultima], amplitudes[primera:ultima], ".-", markersize=4)
plt.title("Ampliación de 20 ms: se distinguen las muestras individuales")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud normalizada")
plt.show()

print("En 20 ms hay", ultima - primera, "muestras.")
```
